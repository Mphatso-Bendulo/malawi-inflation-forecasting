# 01 Data Inspection

This notebook performs an initial systematic inspection of the raw dataset(s) used in the Malawi Inflation Forecasting Lab. It documents dimensions, data types, missing values, duplicate records, and basic statistical summaries. The outputs will inform subsequent data preparation steps.

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
# Load the cleaned inflation data (monthly)
data_path = '../data/cleaned_malawi_inflation_2003_2026.csv'
df = pd.read_csv(data_path)
df.head()
```

```python
print('Shape:', df.shape)
print('Column Types:')
print(df.dtypes)
```

```python
# Check for duplicate rows
dup_rows = df.duplicated().sum()
print(f'Duplicate rows: {dup_rows}')
```

```python
# Missing values per column
missing = df.isna().sum()
print('Missing values per column:')
print(missing)
```

```python
# Basic descriptive statistics for numeric columns
print(df.describe())
```
